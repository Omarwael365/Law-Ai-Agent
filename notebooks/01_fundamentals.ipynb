## 1.1 How Transformer-based LLMs Generate Text

LLMs like GPT-4o use the **transformer architecture** with self-attention mechanisms.
Text generation works through **next-token prediction** — the model predicts the most
likely next token given the preceding context.

Key concepts:
- **Tokenization**: Text is split into tokens (subwords)
- **Self-Attention**: Each token attends to all other tokens in the context
- **Autoregressive Generation**: Tokens are generated one at a time, left to right
- **Temperature**: Controls randomness (lower = more deterministic)

## 1.2 Why LLMs Hallucinate and How RAG Addresses This

**Hallucination** occurs when LLMs generate plausible-sounding but factually incorrect information.
This is especially dangerous in legal contexts where citing non-existent cases can have serious consequences.

**RAG (Retrieval-Augmented Generation)** addresses this by:
1. **Retrieving** relevant documents from a trusted knowledge base
2. **Grounding** the LLM's response in the retrieved text
3. **Providing citations** so users can verify the information
4. **Reducing the need** for the LLM to rely on parametric (memorized) knowledge

## 1.3 What Are Embeddings?

Embeddings map text to high-dimensional vectors where **semantically similar texts
are close together** in vector space. For legal text:

- "force majeure clause" and "act of God provision" would have similar embeddings
- "non-compete agreement" and "restrictive covenant" would be nearby
- "contract law" and "quantum physics" would be far apart

```python
import sys
sys.path.insert(0, '..')

from src.embeddings.embeddings import get_embeddings

# Initialize embeddings
embeddings = get_embeddings(provider='huggingface')

# Demonstrate semantic similarity
legal_terms = [
    'force majeure clause',
    'act of God provision',
    'non-compete agreement',
    'restrictive covenant',
    'quantum physics experiment'
]

vectors = [embeddings.embed_query(term) for term in legal_terms]

# Calculate cosine similarity
import numpy as np

def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

print('Semantic Similarity Matrix:')
print(f"{'':30s}", end='')
for term in legal_terms:
    print(f'{term[:15]:>16s}', end='')
print()

for i, term_i in enumerate(legal_terms):
    print(f'{term_i:30s}', end='')
    for j, term_j in enumerate(legal_terms):
        sim = cosine_similarity(vectors[i], vectors[j])
        print(f'{sim:16.3f}', end='')
    print()
```
