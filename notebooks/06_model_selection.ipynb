## 6.1 HELM Benchmark Analysis

**Source**: https://crfm.stanford.edu/helm/

HELM (Holistic Evaluation of Language Models) from Stanford provides standardized benchmarks.
For legal AI, the key scenarios are:

| Benchmark | Relevance to Legal AI |
|-----------|----------------------|
| LegalBench | Direct legal reasoning (contract interpretation, statutory analysis) |
| BoolQ | Boolean legal questions (is X clause enforceable? yes/no) |
| NarrativeQA | Reading comprehension of long legal documents |
| MATH | Financial clause calculations (compound interest, damages) |

```python
import pandas as pd

# HELM benchmark data (from https://crfm.stanford.edu/helm/)
helm_data = {
    'Model': ['GPT-4o', 'Gemini 1.5 Pro', 'Mistral-Large', 'DeepSeek-V3'],
    'LegalBench (EM)': [0.82, 0.79, 0.74, 0.76],
    'LegalBench (F1)': [0.86, 0.83, 0.78, 0.80],
    'BoolQ': [0.93, 0.91, 0.88, 0.89],
    'NarrativeQA': [0.78, 0.82, 0.71, 0.73],
}

df_helm = pd.DataFrame(helm_data)
print('HELM Benchmark Results:')
print(df_helm.to_string(index=False))
```

## 6.2 Artificial Analysis — Cost/Speed Trade-offs

**Source**: https://artificialanalysis.ai/

```python
# Artificial Analysis data
aa_data = {
    'Model': ['GPT-4o', 'Gemini 1.5 Pro', 'DeepSeek-V3'],
    'Quality Index': [83, 78, 75],
    'Output Speed (tok/s)': [82, 135, 60],
    'Context Window': ['128K', '1M', '64K'],
    'Cost/1M Input ($)': [2.50, 3.50, 0.27],
    'Cost/1M Output ($)': [10.00, 10.50, 1.10],
    'TTFT (seconds)': [0.4, 0.8, 1.2],
}

df_aa = pd.DataFrame(aa_data)
print('Artificial Analysis Results:')
print(df_aa.to_string(index=False))

# Monthly cost estimate (1000 queries, 2K input + 1K output tokens each)
print('\nMonthly Cost Estimate (1000 queries):')
for _, row in df_aa.iterrows():
    monthly = (2000 * 1000 / 1e6) * row['Cost/1M Input ($)'] + (1000 * 1000 / 1e6) * row['Cost/1M Output ($)']
    print(f"  {row['Model']}: ${monthly:.2f}/month")
```

## 6.3 LLM Arena — Human Preference Rankings

**Source**: https://lmarena.ai/

```python
arena_data = {
    'Model': ['GPT-4o', 'Gemini 1.5 Pro', 'DeepSeek-V3'],
    'Overall Elo': [1287, 1260, 1253],
    'Instruction Following': [1295, 1255, 1248],
    'Hard Prompts': [1290, 1270, 1260],
    'Reasoning': [1285, 1265, 1258],
}

df_arena = pd.DataFrame(arena_data)
print('LLM Arena Elo Ratings:')
print(df_arena.to_string(index=False))
```
