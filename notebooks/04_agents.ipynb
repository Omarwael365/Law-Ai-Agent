## 4.1 Option A — LangChain ReAct Agent

The ReAct agent uses a Thought-Action-Observation loop:
1. **Thought**: The agent reasons about what information it needs
2. **Action**: It calls a tool (search_case_law, search_contracts, etc.)
3. **Observation**: It receives the tool's output
4. **Repeat** until it has enough information for a final answer

```python
from src.agents.react_agent import run_react_agent

question = 'Is a non-compete clause enforceable for a software engineer in California?'
result = run_react_agent(question)

print('Answer:')
print(result['answer'])
print(f'\nSteps taken: {len(result["steps"])}')
```

## 4.2 Option B — LangGraph Stateful Agent (Recommended)

LangGraph adds:
- **State management** across conversation turns
- **Multi-step reasoning** with decision nodes
- **Sufficiency checking** — retrieves more if context is insufficient

```python
from src.agents.langgraph_agent import run_langgraph_agent

question = 'Can a departing VP be stopped from using customer lists at a competitor? What legal protections exist?'
result = run_langgraph_agent(question)

print('Answer:')
print(result['answer'])
print(f'\nTool calls: {result["tool_calls"]}')
print(f'Total messages: {result["total_messages"]}')
```

## 4.3 Multi-Turn Conversation

The LangGraph agent supports follow-up questions with memory.

```python
from langchain.schema import HumanMessage, AIMessage

# First question
result1 = run_langgraph_agent('What is force majeure?')
print('Q1 Answer:', result1['answer'][:300])

# Follow-up with chat history
chat_history = [
    HumanMessage(content='What is force majeure?'),
    AIMessage(content=result1['answer'])
]

result2 = run_langgraph_agent(
    'Does COVID-19 qualify under such clauses?',
    chat_history=chat_history
)
print('\nQ2 Answer:', result2['answer'][:300])
```
